```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
from sklearn.model_selection import cross_val_score
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import LeaveOneOut
from mpl_toolkits import mplot3d

# import a txt data file with tab as delimiter
data = pd.read_csv('hcw-data-57.txt', sep='\t| ', header=None)
# a shorterned dataset. Because line 58 of the original dataset has 
# some problems when importing with read_csv
```

```python
data.head()
data.shape
```

```python
scalar = StandardScaler()
scalar.fit(data)
X = scalar.transform(data)
```

```python
X.shape
```

```python
# plot the data covariance matrix as heatmap
plt.figure(figsize=(10,10))
sns.heatmap(np.cov(X.T), annot=False, fmt='.2f', cmap='coolwarm')
plt.show()
```
